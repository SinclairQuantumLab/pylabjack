## Hardware test input template

Replace every all-capital placeholder before running this notebook. Values are passed directly to the project enums and LJM so malformed inputs retain their native errors.

- `<DEVICE_TYPE>`: a `LabJackDeviceTypeEnum` member name
- `<CONNECTION_TYPE>`: a `LabJackConnectionTypeEnum` member name
- `<DEVICE_IDENTIFIER>`: the selected device identifier
- `<AIN_CHANNEL_1>`, `<AIN_CHANNEL_2>`: the physically connected AIN names

## Connection

```python
import numpy as np

from pylabjack import (
    LabJackConnectionTypeEnum,
    LabJackDevice,
    LabJackDeviceTypeEnum,
)

DEVICE_TYPE = "<DEVICE_TYPE>"
CONNECTION_TYPE = "<CONNECTION_TYPE>"
DEVICE_IDENTIFIER = "<DEVICE_IDENTIFIER>"
INPUT_CHANNELS = ["<AIN_CHANNEL_1>", "<AIN_CHANNEL_2>"]

lj_device = LabJackDevice(
    device_type=LabJackDeviceTypeEnum[DEVICE_TYPE],
    connection_type=LabJackConnectionTypeEnum[CONNECTION_TYPE],
    device_identifier=DEVICE_IDENTIFIER,
)
```

## Stream

### Stream In

```python
stream = lj_device.stream(
    scan_rate_Hz=25e3,
    duration_s=10,
    inputs={"channels": INPUT_CHANNELS},
)

print(stream)
print()
total_nans = np.sum([np.isnan(value['V']).sum() for value in stream.records.values()])
print(f"Recounting skipped total samples = {total_nans}")
lj_device.close()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
n_skip_head = 100
for channel, record in stream.records.items():
    ax.plot(record["t"][n_skip_head:], record["V"][n_skip_head:], label=channel)

    
```
